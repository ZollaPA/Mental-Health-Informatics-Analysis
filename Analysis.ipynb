```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create a simulated dataset based on your thesis themes
data = {
    'Region': ['Northeast', 'South', 'Midwest', 'West', 'South', 'Northeast', 'West', 'Midwest'],
    'Wait_Time_Days': [12, 35, 18, 42, 29, 10, 38, 22],
    'Access_Barrier': ['Cost', 'Provider Gap', 'Stigma', 'Provider Gap', 'Cost', 'Cost', 'Geography', 'Stigma'],
    'Insurance_Type': ['Private', 'Uninsured', 'Public', 'Uninsured', 'Public', 'Private', 'Uninsured', 'Public']
}

df = pd.DataFrame(data)

# 2. Analyze the top barriers to access
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Access_Barrier', palette='viridis')
plt.title('Top Barriers to Mental Health Access (Simulated Data)')
plt.xlabel('Barrier Type')
plt.ylabel('Patient Count')
plt.show()

# 3. Analyze Wait Times by Region (The Informatics Problem)
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Region', y='Wait_Time_Days', hue='Insurance_Type')
plt.axhline(14, ls='--', color='red', label='Clinical Benchmark (14 Days)')
plt.title('Average Wait Times for Mental Health Consults by Region')
plt.legend()
plt.show()

# 4. Display the raw data summary
print("--- Informatics Gap Analysis Summary ---")
print(df.groupby('Access_Barrier')['Wait_Time_Days'].mean())
```
